# Image Simulator

This notebook is used for the creation of simulated LSST Camera exposures containing:
* Read noise
* Sky noise
* Background subtraction
* Streak profiles of varying widths, lengths, orientations, and SNR.

```python
%matplotlib inline
import matplotlib.pyplot as plt
import numpy as np
from scipy.stats import norm
```

```python
# Image parameters
ny, nx = (4004, 4096)
read_noise = 7.0  # electrons
sky_counts = {
    "u": 81.0,
    "g": 411.0,
    "r": 819.0,
    "i": 1173.0,
    "z": 1784.0,
    "Y": 2371.0,
}  # electrons
band = "i"

# Streak parameters
theta = 0.0
rho = 2000.0
sigma = 3.0
snr = 100.0
peak_signal = snr * np.sqrt(sky_counts[band])

# Create streak perpendicular profile
gy, gx = np.ogrid[:ny, :nx]
distance = gx*np.cos(theta) + gy*np.sin(theta) - rho
cross_section = norm(0.0, sigma)
streak_signal = cross_section.pdf(distance.flatten()) * sigma * np.sqrt(2 * np.pi) * peak_signal

# Create image plane
rng = np.random.default_rng()
image = rng.poisson(lam=sky_counts[band], size=(ny, nx)) - sky_counts[band]
image += rng.normal(loc=0.0, scale=read_noise, size=(ny, nx))
image += rng.poisson(streak_signal, ny*nx).reshape((ny, nx))

# Create variance plane
variance = np.full((ny, nx), read_noise**2)
variance += sky_counts[band]
variance += streak_signal.reshape((ny, nx))

fig, ax = plt.subplots(1, 1)
im = ax.imshow(image, origin="lower")
cbar = fig.colorbar(im)
```
