# ADRT Detector

This notebook is for the development and testing of the ADRT detector.

1. Array preparation.
2. Match-filter with PSF.
3. Perform ADRT
4. Perform peak finding/analysis.

```python
%matplotlib inline
import matplotlib.pyplot as plt
import numpy as np

import adrt
from lsst.daf.butler import Butler
from lsst.obs.lsst import LsstCam
from mixcoatl.streaks.detectStreaks import get_bad_pixel_mask
```

```python
# Get example
repo = '/sdf/group/rubin/repo/dp2_prep'
camera = LsstCam().getCamera()
instrument = camera.getName()
collections = 'u/snyder18/DRP/20250515-20251214/clusterStreaks'

butler = Butler(repo, collections=collections, instrument=instrument)

visit, detector_id = (2025071700653, 60)
pvi = butler.get("preliminary_visit_image", instrument=instrument, visit=visit, detector=detector_id)
diffim = butler.get("difference_image", instrument=instrument, visit=visit, detector=detector_id)
```

```python
exposure = diffim
image = exposure.image.array
variance = exposure.variance.array
invalid = get_bad_pixel_mask(exposure, ["NO_DATA", "INTRP", "BAD", "SAT", "EDGE", "ITL_DIP", "SPIKE"], 8)

weighted_image = image / variance
weighted_image[invalid] = 0.0
weighted_image = np.pad(weighted_image, ((0, 4096 - image.shape[0]), (0, 0)), mode="constant", constant_values=0.0)

fig, ax = plt.subplots(1, 1)
im = ax.imshow(weighted_image, origin="lower")
cbar = fig.colorbar(im)
```

```python
adrt_result = adrt.adrt(weighted_image)

# Display result
fig, axes = plt.subplots(1, 4, figsize=(10, 8), facecolor="white", sharey=True)
for i, ax in enumerate(axes.ravel()):
    im_plot = ax.imshow(adrt_result[i], vmin=0, vmax=np.max(adrt_result))
fig.tight_layout()
fig.colorbar(im_plot, ax=axes, orientation="horizontal")
```

## Peak Detection and Analysis

```python
q, h, s = np.unravel_index(np.argmax(adrt_result), adrt_result.shape)
print(q, h, s)

offset, angle = adrt.utils.coord_adrt(4096)

print(offset[q, h, s] * 4096)
print(angle[q, 0, s] * 180 / np.pi)
```

```python
peak_region = adrt_result[q, h-128:h+128, s:s+64]
peak_region[peak_region < 0] = 0.0
values = peak_region
positions = np.arange(256)

M0 = np.sum(values, axis=0)
print(M0, M0.shape)
M1 = np.sum(values * positions[:, None], axis=0) / M0
print(M1)
M2 = np.sum(values * (positions[:, None] - M1[None, :])**2, axis=0) / M0
print(M2)
```

```python
slope, intercept = np.polyfit(np.arange(M1.size), M1, 1)
print(slope, intercept)
```
