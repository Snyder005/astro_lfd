# Approximate Discrete Radon Transform

The forward Radon transform computes integrals along lines in an input image (NumPy array) at various angles and offsets. The approximate discrete Radon transform (ADRT) is a discretization of the transform that computes integrals over pixel line segments, which allows for a faster evaluation over large, unbinned images. The ADRT output is a set of 4 arrays in slope-height coordinates, each corresponding to an angle interval of $\frac{\pi}{4}$ for the evaluated line. This "accumulator space" aggregates all the line integral results into slope-height "cells" where the cell value represents the "detection significance" of the linear feature.

Each peak in accumulator space corresponds to a line parameterized by $\rho$, $\theta$ in the image. To extract information about the length and width of the line requires the inverse transform from accumulator space to pixel space, which can introduce discretization errors. Instead, it is preferable to extract additional information about the detected line directly from the accumulator space. For Hough-based transforms a closed-form solution exists to extract the length $L$, width $W$, center point $(x_c, y_c)$ and end points $p_0$, $p_1$. Do similar solutions or estimators exist in ADRT($s$, $h$) accumulator space?

Progess So Far:
1. The accumulator space is $ADRT_{ij}$ where $i$,$j$ are the slope, height indices respectively. Note: in NumPy arrays the index order is $[j, i]$. Let $f(\vec{x})$ be a parameterization of a line with slope $s_0$ and height $h_0$.
2. Let $Q_i$ be the number of nonzero bins in the $i$-th accumulator column. Then $$Q_i \approx \frac{L|\cos(\theta_0)|}{N} |s_0 - s_i| + W,$$ where $W$ is the line segment width and $L$ is the line segment length. For $|\theta_0| > \frac{\pi}{4}$, the $\cos(\theta)$ term becomes $\sin(\theta)$, as the line orientation changes.
3. The summed value in the $i$-th accumulator column is $$A = \sum_j ADRT(s_0, h_j) = \int_L f(\vec{x}) d\vec{x}$$ where $L$ is the length of the line. For a binary image, $$A = L|\cos(\theta_0)|$$
4. The centroid of the $s_i$ accumulator column is the height cell $c_i$. Then $$c_i \approx \frac{D|\cos(\theta_0)|}{N} |s_0 - s_i| + b_0,$$ where $D$ is the distance of the line segment center point, along the line, from the left image boundary (where $h$ is defined for the "digital" extended line). Again the trigonometric function changes as before depending on line orientation ($\theta_0$).

General Observations:
1. For $s_0$ the column profile is approximately a 1-D perpendicular profile of the line in $Q_0$ bins. Thus $Q_0 \approx W$. Note: this is an integer estimate of the line width.
2. For $s_i$ far from $s_0$ the column profile approaches a 1-D parallel profile along the line. Thus $ADRT(s_i, h) \to f(\vec{x})$. As $i$ increases the number of bins $Q_i$ increases, improving the fidelity of the profile.

Next Steps:

Unlike voting distribution in Hough space, which is a discrete random variable in a noisy image, the accumulator wedge contains information about the signal distribution of the line.
1. For grayscale (`float`) images, instead study the per-column moments (i.e., $\mu_i$, $\sigma_i$) as estimators of $Q_i$ and $s_i$.
2. Entropy in the column: $$p_i(h) = \frac{A(s_i, h)}{\sum_h A(s_i, h)}.$$ Then calculate: $$H_i = - \sum_h p_i(h) \log p_i(h).$$

```python
%matplotlib inline
import adrt
import matplotlib.pyplot as plt
import numpy as np
from skimage.draw import line

def get_rho_theta(q, h, s):
    offset, angle = adrt.utils.coord_adrt(N)
    xc = (N - 1) / 2
    yc = (N - 1) / 2
    
    theta = np.pi / 2 - angle[q, 0, s]
    rho = xc * np.cos(theta) + yc * np.sin(theta) - offset[q, h, s] * N

    return rho, theta

#peak = np.unravel_index(np.argmax(adrt_result), adrt_result.shape) 
#q, h, s = peak
```

# Accumulator Column Profiles

```python
from numpy.polynomial import Polynomial

N = 2048
imarr = np.zeros((N, N))
imarr[:, 400:420] += 1.0


fig, ax = plt.subplots(1, 1, figsize=(10, 8), facecolor="white")
im = ax.imshow(imarr)
cbar = fig.colorbar(im)
plt.show()
```

```python
# Get the ADRT result
adrt_result = adrt.adrt(imarr)

# Display result
fig, axs = plt.subplots(1, 4, figsize=(10, 8), facecolor="white", sharey=True)
for i, ax in enumerate(axs.ravel()):
    im_plot = ax.imshow(adrt_result[i], vmin=0, vmax=np.max(adrt_result))
fig.tight_layout()
fig.colorbar(im_plot, ax=axs, orientation="horizontal")
```

```python
adrt_stitched = adrt.utils.stitch_adrt(adrt_result)

# Display result
fig, ax = plt.subplots(1, 1, figsize=(10, 8), facecolor="white")
im = ax.imshow(adrt_stitched)
cbar = fig.colorbar(im)
for i in range(1, 4):
    ax.axvline(N * i - 0.5, color="white", linestyle="--")
fig.tight_layout()
```

```python
# Get peak (s, h) and theta
peak = np.unravel_index(np.argmax(adrt_result), adrt_result.shape)
q, h, s = peak
print(q, h, s)
offset, angle = adrt.utils.coord_adrt(N)
theta = angle[q, 0, s]
print(theta)
if np.abs(theta) > np.pi / 4:
    print(np.abs(np.cos(np.pi / 2 - theta)))
    print(np.abs(np.sin(theta)))

# Get a region containing the peak
peak_region = adrt_result[q, h:h+256, s:256]
fig, ax = plt.subplots(1, 1, figsize=(10, 8), facecolor="white")
ax.imshow(peak_region)
```

```python
# Get cell width by column
cell_width = np.count_nonzero(peak_region, axis=0)
print(f"Line Width = {cell_width[0]:.1f}") # 0-th column is line width

# Linear fit to get to get width and length
x = np.arange(cell_width.size)
line_fit = Polynomial.fit(x, cell_width, deg=1) # adding higher degrees may remove discretization effects
intercept, slope = line_fit.convert().coef
print(f"Line Width = {intercept:.2f}, Line Length = {slope * N / np.sin(-theta):.2f}")

# Calculate cell column center 
first_nonzero_indices = (peak_region != 0).argmax(axis=0)
last_indices = np.where(
    np.count_nonzero(peak_region, axis=0) == 0, 
    -1, 
    (peak_region.shape[0] - 1) - np.argmax(peak_region[::-1, :] != 0, axis=0),
)
center_idx = (last_indices + first_nonzero_indices) / 2.0

# Linear fit to get width/2 and distance of line segment center along the line
line_fit = Polynomial.fit(x, center_idx, deg=1)
intercept, slope = line_fit.convert().coef
print(f"Line Width = {2 * intercept:.2f}, Line Center Distance = {slope * N:.2f}")
```

```python
print(np.mean(np.sum(peak_region, axis=0)))
```

Observations:
1. Binary/Float Images: The width of the accumulator column is the pixel length of the line times the slope index offset minus the line width. Can estimate from a linear fit. $$w(s) = \frac{L}{N}s + W$$
2. Binary/Float Images: The center of the accumulator column is the distance of the center point along the line times the slope index offset minus half the line width. Can estimate from a linear fit. $$c(s) = \frac{D}{N}s + \frac{W}{2}$$
3. Binary/Float Images: The sum of the accumulator column (within a "maximum" triangle wedge at 45 degrees) is equal to the integral of the line.
    1. Binary Images: This is exactly the pixel length of the line.


```python
fig, ax = plt.subplots(1, 1, figsize=(10, 8), facecolor="white")
for i in range(4):
    ax.plot(peak_region[:200, 64*i], label=f"i={64*i}")

ax.legend(fontsize=10)
```

## Noise Tests

```python
N = 2048
sky = 500
noise = 7.0
var = sky + noise**2
sigma = np.sqrt(var)
imarr = np.zeros((N, N))
imarr += np.random.normal(0.0, noise, imarr.shape)
imarr += (np.random.poisson(lam=sky, size=imarr.shape) - sky)
imarr[100:120, 400:1200] += 0.5 * sigma
imarr /= np.sqrt(sigma)

#sigma = np.sqrt(sky)
print(sigma)
print(100 / sigma)
#imarr[400, :] = 100 * sigma

fig, ax = plt.subplots(1, 1, figsize=(10, 8), facecolor="white")
im = ax.imshow(imarr)
cbar = fig.colorbar(im)
plt.show()
```

```python
# Get the ADRT result
adrt_result = adrt.adrt(imarr)
print(adrt_result.dtype)

# Display result
fig, axs = plt.subplots(1, 4, figsize=(10, 8), facecolor="white", sharey=True)
for i, ax in enumerate(axs.ravel()):
    im_plot = ax.imshow(adrt_result[i], vmin=0, vmax=np.max(adrt_result))
fig.tight_layout()
fig.colorbar(im_plot, ax=axs, orientation="horizontal")
```

```python
# Get peak (s, h) and theta
peak = np.unravel_index(np.argmax(adrt_result), adrt_result.shape)
q, h, s = peak
print(q, h, s)
offset, angle = adrt.utils.coord_adrt(N)
theta = angle[q, 0, s]

# Get a region containing the peak
peak_region = adrt_result[q, h-100:h+256, s:64]
fig, ax = plt.subplots(1, 1, figsize=(10, 8), facecolor="white")
ax.imshow(peak_region, interpolation="none")
```

```python
# Get cell width by column
cell_width = np.count_nonzero(peak_region > 1000.0, axis=0)
print(cell_width)
print(f"Line Width = {cell_width[0]:.1f}") # 0-th column is line width

# Linear fit to get to get width and length
x = np.arange(cell_width.size)
line_fit = Polynomial.fit(x, cell_width, deg=2) # adding higher degrees may remove discretization effects
intercept, slope, _ = line_fit.convert().coef
print(f"Line Width = {intercept:.2f}, Line Length = {slope * N:.2f}")

# Calculate cell column center 
first_nonzero_indices = (peak_region > 1000.0).argmax(axis=0)
last_indices = np.where(
    np.count_nonzero(peak_region > 1000.0, axis=0) == 0,
    -1, 
    (peak_region.shape[0] - 1) - np.argmax(peak_region[::-1, :] != 0, axis=0),
)
center_idx = (last_indices + first_nonzero_indices) / 2.0
print((last_indices - first_nonzero_indices))

# Linear fit to get width/2 and distance of line segment center along the line
line_fit = Polynomial.fit(x, center_idx, deg=1)
intercept, slope = line_fit.convert().coef
print(f"Line Width = {2 * intercept:.2f}, Line Center Distance = {slope * N:.2f}")
```
